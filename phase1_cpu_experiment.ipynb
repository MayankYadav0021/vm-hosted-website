```python
# Cell 2: Production-Grade Monitoring with Real-World Metrics
import time
from datetime import datetime
import psutil
import torch

class ProductionMonitor:
    def __init__(self):
        self.metrics_history = []
        self.start_time = time.time()
        self.performance_baselines = {}

    def collect_comprehensive_metrics(self, phase="", model_name="", inference_time=0):
        """Collect production-grade metrics"""
        metrics = {
            'timestamp': time.time() - self.start_time,
            'phase': phase,
            'model': model_name,
            'datetime': datetime.now(),
            'inference_time': inference_time
        }

        # GPU Metrics
        if torch.cuda.is_available():
            metrics.update({
                'gpu_memory_allocated_gb': torch.cuda.memory_allocated() / 1e9,
                'gpu_memory_reserved_gb': torch.cuda.memory_reserved() / 1e9,
                'gpu_utilization_estimated': min(100, (torch.cuda.memory_allocated() / torch.cuda.get_device_properties(0).total_memory) * 100),
            })

            # Simulate additional GPU metrics (real NVML would be used in production)
            torch.cuda.synchronize()

        # System Metrics
        cpu_percent = psutil.cpu_percent(interval=0.1)
        memory = psutil.virtual_memory()
        disk = psutil.disk_usage('/')

        metrics.update({
            'cpu_percent': cpu_percent,
            'ram_used_gb': memory.used / 1e9,
            'ram_percent': memory.percent,
            'disk_used_percent': disk.percent,
            'system_load': psutil.getloadavg()[0] if hasattr(psutil, 'getloadavg') else cpu_percent / 100,
        })

        # Derived Metrics for Failure Detection
        if inference_time > 0:
            metrics.update({
                'throughput_tokens_per_sec': 100 / inference_time,  # Assuming ~100 tokens per inference
                'latency_per_token_ms': (inference_time * 1000) / 100,
                'efficiency_score': 100 / (inference_time * metrics.get('gpu_memory_allocated_gb', 1)),
            })

        self.metrics_history.append(metrics)
        return metrics

    def calculate_performance_baselines(self):
        """Calculate baselines for anomaly detection"""
        df = pd.DataFrame(self.metrics_history)

        self.performance_baselines = {
            'inference_time_mean': df['inference_time'].mean(),
            'inference_time_std': df['inference_time'].std(),
            'memory_usage_mean': df['gpu_memory_allocated_gb'].mean(),
            'memory_usage_std': df['gpu_memory_allocated_gb'].std(),
            'throughput_mean': df['throughput_tokens_per_sec'].mean(),
            'throughput_std': df['throughput_tokens_per_sec'].std(),
        }

        return self.performance_baselines

# Initialize production monitor
production_monitor = ProductionMonitor()
```

```python
# =============================================================================
# FIXED COMBINED LIGHTWEIGHT MODELS - DEVICE ISSUES RESOLVED
# =============================================================================

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import logging
from typing import Dict, Any, List
import time
import threading
from concurrent.futures import ThreadPoolExecutor
from dataclasses import dataclass
import gc

@dataclass
class ModelWorker:
    model_name: str
    model: Any
    tokenizer: Any
    device: str
    is_busy: bool = False
    last_used: float = 0

class FixedLightweightLoadBalancer:
    def __init__(self):
        self.available_models = {
            "tiny-llama": "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
            "phi-2": "microsoft/phi-2",
            "qwen-1.8b": "Qwen/Qwen1.5-1.8B-Chat",
        }
        self.model_workers: Dict[str, List[ModelWorker]] = {}
        self.worker_lock = threading.Lock()
        self.executor = ThreadPoolExecutor(max_workers=3)
        self.setup_logging()

    def setup_logging(self):
        logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(message)s')
        self.logger = logging.getLogger(__name__)

    def clear_memory(self):
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    def load_model_with_fallback(self, model_name: str, model_path: str):
        """Load model with proper device handling"""
        try:
            self.logger.info(f"🔄 Loading {model_name}...")
            self.clear_memory()

            # Load tokenizer first
            tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)
            if tokenizer.pad_token is None:
                tokenizer.pad_token = tokenizer.eos_token

            # Try loading with different strategies
            try:
                # Strategy 1: With quantization
                quantization_config = BitsAndBytesConfig(
                    load_in_4bit=True,
                    bnb_4bit_use_double_quant=True,
                    bnb_4bit_quant_type="nf4",
                    bnb_4bit_compute_dtype=torch.float16
                )

                model = AutoModelForCausalLM.from_pretrained(
                    model_path,
                    quantization_config=quantization_config,
                    device_map="auto",
                    trust_remote_code=True,
                    torch_dtype=torch.float16
                )
            except Exception as quant_error:
                self.logger.warning(f"Quantization failed, loading normally: {quant_error}")
                # Strategy 2: Without quantization
                model = AutoModelForCausalLM.from_pretrained(
                    model_path,
                    device_map="auto",
                    trust_remote_code=True,
                    torch_dtype=torch.float16
                )

            # Get device info
            device = str(next(model.parameters()).device)

            worker = ModelWorker(
                model_name=model_name,
                model=model,
                tokenizer=tokenizer,
                device=device
            )

            self.logger.info(f"✅ {model_name} loaded on {device}")
            return worker

        except Exception as e:
            self.logger.error(f"❌ Failed to load {model_name}: {e}")
            return None

    def initialize_all_models(self):
        """Initialize models with error handling"""
        self.logger.info("🚀 Initializing lightweight models...")

        successful_models = 0

        for model_name, model_path in self.available_models.items():
            worker = self.load_model_with_fallback(model_name, model_path)
            if worker:
                self.model_workers[model_name] = [worker]
                successful_models += 1
                print(f"✅ {model_name}: Loaded on {worker.device}")
            else:
                print(f"❌ {model_name}: Failed to load")

        print(f"🎯 {successful_models}/3 models loaded successfully")
        return successful_models > 0

    def submit_request(self, prompt: str, model_name: str = None) -> str:
        """Process request with proper device handling"""
        if not self.model_workers:
            return "Error: No models loaded."

        available_models = list(self.model_workers.keys())

        if model_name is None:
            model_name = available_models[0]

        if model_name not in available_models:
            return f"Error: Model {model_name} not available. Choose from: {available_models}"

        worker = self.model_workers[model_name][0]

        if worker.is_busy:
            return "Error: Model busy, please wait..."

        worker.is_busy = True
        worker.last_used = time.time()

        try:
            self.clear_memory()

            # Model-specific prompt formatting
            if model_name == "tiny-llama":
                formatted_prompt = f"<|user|>\n{prompt}</s>\n<|assistant|>\n"
            elif model_name == "phi-2":
                formatted_prompt = f"Instruct: {prompt}\nOutput:"
            elif model_name == "qwen-1.8b":
                formatted_prompt = f"<|im_start|>user\n{prompt}<|im_end|>\n<|im_start|>assistant\n"
            else:
                formatted_prompt = prompt

            # Tokenize and move to correct device
            inputs = worker.tokenizer(formatted_prompt, return_tensors="pt", truncation=True, max_length=1024)

            # FIX: Move inputs to the same device as model
            model_device = next(worker.model.parameters()).device
            inputs = {key: value.to(model_device) for key, value in inputs.items()}

            # Generate response
            with torch.no_grad():
                outputs = worker.model.generate(
                    **inputs,
                    max_new_tokens=150,
                    temperature=0.7,
                    do_sample=True,
                    pad_token_id=worker.tokenizer.eos_token_id
                )

            # Decode response
            response = worker.tokenizer.decode(outputs[0], skip_special_tokens=True)

            # Clean response based on model
            if model_name == "tiny-llama":
                response = response.split("<|assistant|>")[-1].strip()
            elif model_name == "phi-2":
                response = response.split("Output:")[-1].strip()
            elif model_name == "qwen-1.8b":
                response = response.split("<|im_start|>assistant")[-1].replace("<|im_end|>", "").strip()

            return response

        except Exception as e:
            return f"Error: {str(e)}"
        finally:
            worker.is_busy = False
            self.clear_memory()

# =============================================================================
# SIMPLIFIED CHAT INTERFACE
# =============================================================================

def fixed_chat_interface():
    """Fixed chat interface without complex features"""

    print("🤖 FIXED AI CHAT SYSTEM")
    print("Loading models...")

    # Initialize load balancer
    load_balancer = FixedLightweightLoadBalancer()

    if not load_balancer.initialize_all_models():
        print("❌ No models loaded. Please check errors above.")
        return

    available_models = list(load_balancer.model_workers.keys())
    current_model = available_models[0]

    print(f"\n💡 TIPS:")
    print("• Just type your message and press Enter")
    print("• Type 'exit' to quit")
    print("• Type 'switch' to change model")
    print(f"• Current model: {current_model}")

    while True:
        try:
            user_input = input(f"\n[{current_model}] You: ").strip()

            if user_input.lower() in ['exit', 'quit']:
                break
            elif user_input.lower() == 'switch':
                print(f"Available models: {', '.join(available_models)}")
                new_model = input("Enter model name: ").strip()
                if new_model in available_models:
                    current_model = new_model
                    print(f"✅ Switched to {current_model}")
                else:
                    print(f"❌ Invalid model. Available: {available_models}")
                continue
            elif user_input.lower() == '':
                continue

            print("⏳ Processing...")
            start_time = time.time()

            response = load_balancer.submit_request(user_input, current_model)

            processing_time = time.time() - start_time
            print(f"🤖 Assistant: {response}")
            print(f"⏱️  Time: {processing_time:.2f}s")

        except KeyboardInterrupt:
            print("\n👋 Goodbye!")
            break
        except Exception as e:
            print(f"❌ Unexpected error: {e}")

# =============================================================================
# QUICK TEST FUNCTION
# =============================================================================

def quick_test():
    """Quick test to verify everything works"""

    print("🧪 QUICK SYSTEM TEST")
    print("=" * 40)

    load_balancer = FixedLightweightLoadBalancer()

    if load_balancer.initialize_all_models():
        test_prompt = "Hello, how are you today?"

        for model_name in load_balancer.model_workers.keys():
            print(f"\n🔹 Testing {model_name}...")
            start_time = time.time()

            response = load_balancer.submit_request(test_prompt, model_name)

            processing_time = time.time() - start_time
            print(f"✅ Response: {response}")
            print(f"⏱️  Time: {processing_time:.2f}s")

        print("\n🎉 All tests completed successfully!")
    else:
        print("❌ System test failed - models not loaded")

# =============================================================================
# MAIN EXECUTION
# =============================================================================

if __name__ == "__main__":
    print("🚀 FIXED LIGHTWEIGHT AI CHAT")
    print("=" * 50)

    # First, clear any existing memory
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    print("Choose option:")
    print("1. Start Chat (Recommended)")
    print("2. Quick Test")

    choice = input("Enter 1 or 2: ").strip()

    if choice == "2":
        quick_test()
    else:
        fixed_chat_interface()
```

```python
# Cell 3: Download Diverse Model Portfolio for Comparative Analysis
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoModelForSeq2SeqLM
import gc
import torch
import time

# Curated model portfolio for comprehensive comparison
model_portfolio = {
    'tiny': [
        "microsoft/DialoGPT-small",      # 80M - Conversational
        "distilbert/distilgpt2",         # 82M - General purpose
    ],
    'small': [
        "microsoft/DialoGPT-medium",     # 300M - Conversational
        "microsoft/DialoGPT-large",      # 762M - Conversational
    ],
    'medium': [
        "facebook/blenderbot-400M-distill", # 400M - Dialog
        "google/pegasus-xsum",           # 568M - Summarization
    ]
}

print("📥 DOWNLOADING DIVERSE MODEL PORTFOLIO")
print("=" * 50)

models_dict = {}
download_metrics = []

for size_category, models in model_portfolio.items():
    print(f"\n🔍 Processing {size_category.upper()} models:")

    for model_name in models:
        print(f"   📦 {model_name}")

        try:
            # Clear memory
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
                torch.cuda.synchronize()

            memory_before = torch.cuda.memory_allocated() / 1e9
            download_start = time.time()

            # Download tokenizer
            tokenizer = AutoTokenizer.from_pretrained(model_name)
            if tokenizer.pad_token is None:
                tokenizer.pad_token = tokenizer.eos_token

            # Download model with appropriate settings
            if "pegasus" in model_name or "blenderbot" in model_name:
                model = AutoModelForSeq2SeqLM.from_pretrained(
                    model_name,
                    torch_dtype=torch.float16,
                    device_map="auto",
                    low_cpu_mem_usage=True
                )
            else:
                model = AutoModelForCausalLM.from_pretrained(
                    model_name,
                    torch_dtype=torch.float16,
                    device_map="auto",
                    low_cpu_mem_usage=True
                )

            download_time = time.time() - download_start
            memory_after = torch.cuda.memory_allocated() / 1e9
            memory_used = memory_after - memory_before

            # Calculate model statistics
            total_params = sum(p.numel() for p in model.parameters())
            model_size_mb = total_params * 2 / 1e6  # Approximate size in MB (assuming float16)

            models_dict[model_name] = {
                'model': model,
                'tokenizer': tokenizer,
                'parameters': total_params,
                'size_category': size_category,
                'architecture': 'seq2seq' if "pegasus" in model_name or "blenderbot" in model_name else 'causal',
                'memory_footprint_gb': memory_used
            }

            download_metrics.append({
                'model': model_name,
                'size_category': size_category,
                'download_time': download_time,
                'memory_used': memory_used,
                'parameters': total_params,
                'model_size_mb': model_size_mb
            })

            production_monitor.collect_comprehensive_metrics(f"download_{size_category}", model_name)

            print(f"      ✅ Parameters: {total_params/1e6:.1f}M | Memory: {memory_used:.2f}GB | Time: {download_time:.1f}s")

        except Exception as e:
            print(f"      ❌ Failed: {e}")
            continue

        # Prevent rate limiting
        time.sleep(2)

print(f"\n🎯 Successfully downloaded {len(models_dict)} models across {len(set([m['size_category'] for m in download_metrics]))} size categories")
```

```python
# Cell 5: Real-World Workload Simulation with Multiple Scenarios
print("🌍 REAL-WORLD WORKLOAD SIMULATION")
print("=" * 50)

class WorkloadSimulator:
    def __init__(self, monitor):
        self.monitor = monitor
        self.workload_results = {}

    def simulate_workload_scenario(self, scenario_name, model, tokenizer, **kwargs):
        """Simulate different real-world workload scenarios"""

        print(f"\n🔄 Simulating: {scenario_name}")

        scenarios = {
            'chat_conversation': {
                'prompts': [
                    "Hello, how are you today?",
                    "What can you tell me about artificial intelligence?",
                    "How does machine learning work?",
                    "What are the benefits of renewable energy?",
                    "Can you explain quantum computing?"
                ],
                'max_length': 100,
                'temperature': 0.7
            },
            'content_generation': {
                'prompts': [
                    "Write a short story about a robot learning to paint:",
                    "Explain the concept of neural networks in simple terms:",
                    "Describe the impact of climate change on coastal cities:",
                    "Create a product description for a new smartphone:"
                ],
                'max_length': 150,
                'temperature': 0.8
            },
            'technical_qa': {
                'prompts': [
                    "What is the difference between supervised and unsupervised learning?",
                    "How does attention mechanism work in transformers?",
                    "Explain the concept of backpropagation in neural networks:",
                    "What are the advantages of using GPU for deep learning?"
                ],
                'max_length': 120,
                'temperature': 0.6
            }
        }

        scenario_config = scenarios.get(scenario_name, scenarios['chat_conversation'])

        performance_metrics = {
            'scenario': scenario_name,
            'total_requests': len(scenario_config['prompts']),
            'inference_times': [],
            'memory_usage': [],
            'throughput': [],
            'generated_lengths': []
        }

        for i, prompt in enumerate(scenario_config['prompts']):
            # Clear cache between requests (simulating real production)
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
                torch.cuda.synchronize()

            memory_before = torch.cuda.memory_allocated() / 1e9
            start_time = time.time()

            inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512).to(model.device)

            with torch.no_grad():
                outputs = model.generate(
                    inputs.input_ids,
                    max_length=scenario_config['max_length'],
                    temperature=scenario_config['temperature'],
                    do_sample=True,
                    pad_token_id=tokenizer.eos_token_id,
                    attention_mask=inputs.attention_mask
                )

            if torch.cuda.is_available():
                torch.cuda.synchronize()

            end_time = time.time()
            memory_after = torch.cuda.memory_allocated() / 1e9

            inference_time = end_time - start_time
            memory_used = memory_after - memory_before

            # Calculate tokens and throughput
            generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
            input_tokens = len(tokenizer.encode(prompt))
            output_tokens = len(tokenizer.encode(generated_text))
            total_tokens = output_tokens - input_tokens
            tokens_per_sec = total_tokens / inference_time if inference_time > 0 else 0

            performance_metrics['inference_times'].append(inference_time)
            performance_metrics['memory_usage'].append(memory_used)
            performance_metrics['throughput'].append(tokens_per_sec)
            performance_metrics['generated_lengths'].append(len(generated_text.split()))

            # Record metrics
            self.monitor.collect_comprehensive_metrics(
                f"{scenario_name}_request_{i}",
                model.config.name_or_path if hasattr(model.config, 'name_or_path') else "unknown",
                inference_time
            )

            if i < 2:  # Print first few for verification
                print(f"   Request {i+1}: {inference_time:.3f}s, {tokens_per_sec:.1f} tok/s")

        # Calculate summary statistics
        performance_metrics.update({
            'avg_inference_time': np.mean(performance_metrics['inference_times']),
            'std_inference_time': np.std(performance_metrics['inference_times']),
            'avg_throughput': np.mean(performance_metrics['throughput']),
            'avg_memory_usage': np.mean(performance_metrics['memory_usage']),
            'total_tokens_generated': sum([len(tokenizer.encode(p)) for p in scenario_config['prompts']])
        })

        return performance_metrics

    def run_comprehensive_workloads(self, models_dict, scenarios=None):
        """Run multiple workload scenarios across all models"""
        if scenarios is None:
            scenarios = ['chat_conversation', 'content_generation', 'technical_qa']

        all_results = []

        for model_name, model_data in models_dict.items():
            short_name = model_name.split('/')[-1]
            print(f"\n🎯 Testing {short_name} across {len(scenarios)} scenarios")

            model_results = {
                'model': short_name,
                'parameters': model_data['parameters'],
                'size_category': model_data['size_category'],
                'architecture': model_data['architecture'],
                'scenarios': {}
            }

            for scenario in scenarios:
                try:
                    scenario_results = self.simulate_workload_scenario(
                        scenario,
                        model_data['model'],
                        model_data['tokenizer']
                    )
                    model_results['scenarios'][scenario] = scenario_results

                    print(f"   {scenario}: {scenario_results['avg_throughput']:.1f} tok/s")

                except Exception as e:
                    print(f"   ❌ {scenario} failed: {e}")
                    continue

            all_results.append(model_results)

        self.workload_results = all_results
        return all_results

# Initialize and run workload simulator
workload_simulator = WorkloadSimulator(production_monitor)

if models_dict:
    workload_results = workload_simulator.run_comprehensive_workloads(models_dict)

    print(f"\n✅ Workload simulation completed: {len(workload_results)} models × 3 scenarios")
```

```python
# Cell 7: Comprehensive Quantization Effects Analysis
print("🔢 COMPREHENSIVE QUANTIZATION IMPACT ANALYSIS")
print("=" * 50)

import torch

class QuantizationAnalyzer:
    def __init__(self, monitor):
        self.monitor = monitor
        self.quantization_results = {}

    def apply_quantization(self, model, quantization_type='fp16'):
        """Apply different quantization techniques"""
        original_dtype = next(model.parameters()).dtype

        try:
            if quantization_type == 'fp32':
                quantized_model = model.to(torch.float32)
                memory_reduction = 0.0

            elif quantization_type == 'fp16':
                quantized_model = model.to(torch.float16)
                memory_reduction = 0.5  # Theoretical 50% reduction

            else:
                quantized_model = model
                memory_reduction = 0.0

            return quantized_model, memory_reduction

        except Exception as e:
            print(f"      Quantization {quantization_type} failed: {e}")
            return model, 0.0

    def test_quantization_effects(self, model_name, model_data, test_prompt):
        """Test quantization effects on performance and accuracy"""
        original_model = model_data['model']
        tokenizer = model_data['tokenizer']

        print(f"\n🔍 Testing quantization on {model_name.split('/')[-1]}")

        quantization_types = ['fp32', 'fp16']  # Removed int8 and int4

        results = []

        for q_type in quantization_types:
            print(f"   Testing {q_type}...")

            try:
                # Clear cache
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                    torch.cuda.synchronize()

                memory_before = torch.cuda.memory_allocated() / 1e9

                # Apply quantization
                quantized_model, theoretical_reduction = self.apply_quantization(original_model, q_type)

                # Measure memory after quantization
                memory_after = torch.cuda.memory_allocated() / 1e9
                actual_memory_used = memory_after - memory_before

                # Test inference performance
                inputs = tokenizer(test_prompt, return_tensors="pt", truncation=True, max_length=512).to(quantized_model.device)

                # Warm-up run
                with torch.no_grad():
                    _ = quantized_model.generate(
                        inputs.input_ids,
                        max_length=50,
                        num_return_sequences=1,
                        pad_token_id=tokenizer.eos_token_id
                    )

                # Performance test
                start_time = time.time()
                with torch.no_grad():
                    outputs = quantized_model.generate(
                        inputs.input_ids,
                        max_length=100,
                        temperature=0.7,
                        do_sample=True,
                        pad_token_id=tokenizer.eos_token_id,
                        attention_mask=inputs.attention_mask
                    )

                if torch.cuda.is_available():
                    torch.cuda.synchronize()

                end_time = time.time()
                inference_time = end_time - start_time

                # Calculate metrics
                generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
                input_tokens = len(tokenizer.encode(test_prompt))
                output_tokens = len(tokenizer.encode(generated_text))
                total_tokens = output_tokens - input_tokens
                tokens_per_sec = total_tokens / inference_time if inference_time > 0 else 0

                results.append({
                    'quantization': q_type,
                    'inference_time': inference_time,
                    'tokens_per_sec': tokens_per_sec,
                    'memory_used_gb': actual_memory_used,
                    'theoretical_memory_reduction': theoretical_reduction,
                    'output_quality': len(generated_text.split()) / len(test_prompt.split()), # Simple length-based quality
                    'coherence_score': min(1.0, len(generated_text) / 100)  # Simple length-based coherence
                })

                print(f"      {q_type}: {inference_time:.3f}s, {tokens_per_sec:.1f} tok/s, {actual_memory_used:.3f}GB")

            except Exception as e:
                print(f"      ❌ {q_type} test failed: {e}")
                continue

        # Calculate relative performance compared to fp32
        if len(results) >= 2:
            fp32_time = next(r['inference_time'] for r in results if r['quantization'] == 'fp32')
            fp32_quality = next(r['output_quality'] for r in results if r['quantization'] == 'fp32')

            for result in results:
                if fp32_time > 0:
                    result['speed_vs_fp32'] = (fp32_time / result['inference_time'] - 1) * 100
                if fp32_quality > 0:
                    result['quality_vs_fp32'] = (result['output_quality'] / fp32_quality - 1) * 100

        return results

    def run_comprehensive_quantization_analysis(self, models_dict, test_prompt):
        """Run quantization analysis across all models"""
        all_results = {}

        for model_name, model_data in models_dict.items():
            short_name = model_name.split('/')[-1]

            results = self.test_quantization_effects(model_name, model_data, test_prompt)
            if results:
                all_results[model_name] = {
                    'model_info': {
                        'name': short_name,
                        'parameters': model_data['parameters'],
                        'size_category': model_data['size_category'],
                        'architecture': model_data['architecture']
                    },
                    'quantization_results': results
                }

        self.quantization_results = all_results
        return all_results

# Run quantization analysis
quantization_analyzer = QuantizationAnalyzer(production_monitor)
test_prompt = "Explain the concept of machine learning and its applications in modern technology:"

if models_dict:
    quantization_results = quantization_analyzer.run_comprehensive_quantization_analysis(models_dict, test_prompt)

    print(f"\n✅ Quantization analysis completed: {len(quantization_results)} models")
```

```python
# Cell 9: Batch Processing Impact on GPU Utilization
print("🔄 BATCH PROCESSING EFFECTS ANALYSIS")
print("=" * 50)

class BatchProcessingAnalyzer:
    def __init__(self, monitor):
        self.monitor = monitor
        self.batch_results = {}

    def test_batch_processing(self, model, tokenizer, base_prompt, batch_sizes=[1, 2, 4, 8, 16]):
        """Test different batch sizes and their effects"""

        print(f"\n🔍 Testing batch processing on {model.config.name_or_path.split('/')[-1]}")

        batch_performance = []

        for batch_size in batch_sizes:
            print(f"   Batch size: {batch_size}")

            try:
                # Create batch
                prompts = [f"{base_prompt} - variation {i}" for i in range(batch_size)]

                # Clear cache
                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                    torch.cuda.synchronize()

                memory_before = torch.cuda.memory_allocated() / 1e9
                start_time = time.time()

                # Tokenize batch
                inputs = tokenizer(prompts, return_tensors="pt", padding=True, truncation=True, max_length=512).to(model.device)

                # Generate with batch
                with torch.no_grad():
                    outputs = model.generate(
                        inputs.input_ids,
                        attention_mask=inputs.attention_mask,
                        max_length=100,
                        temperature=0.7,
                        do_sample=True,
                        pad_token_id=tokenizer.eos_token_id,
                        num_return_sequences=1
                    )

                if torch.cuda.is_available():
                    torch.cuda.synchronize()

                end_time = time.time()
                memory_after = torch.cuda.memory_allocated() / 1e9

                processing_time = end_time - start_time
                memory_used = memory_after - memory_before

                # Calculate efficiency metrics
                throughput = batch_size / processing_time  # requests per second
                memory_per_request = memory_used / batch_size
                gpu_utilization = min(100, (memory_after / (torch.cuda.get_device_properties(0).total_memory / 1e9)) * 100)

                # Parallel efficiency (compared to single request)
                single_request_time = processing_time / batch_size  # Simplified
                parallel_efficiency = (1 / single_request_time) / (batch_size / processing_time) * 100

                batch_performance.append({
                    'batch_size': batch_size,
                    'processing_time': processing_time,
                    'throughput': throughput,
                    'memory_used': memory_used,
                    'memory_per_request': memory_per_request,
                    'gpu_utilization': gpu_utilization,
                    'parallel_efficiency': parallel_efficiency,
                    'time_per_request': processing_time / batch_size
                })

                print(f"      Time: {processing_time:.3f}s, Throughput: {throughput:.2f} req/s, "
                      f"Memory: {memory_used:.3f}GB, GPU Util: {gpu_utilization:.1f}%")

                # Record metrics
                self.monitor.collect_comprehensive_metrics(
                    f"batch_{batch_size}",
                    model.config.name_or_path if hasattr(model.config, 'name_or_path') else "unknown",
                    processing_time
                )

                # Stop if approaching memory limits
                if gpu_utilization > 80:
                    print("      ⚠️  High GPU utilization, stopping batch size increase")
                    break

            except RuntimeError as e:
                if "out of memory" in str(e):
                    print(f"      ❌ OOM at batch size {batch_size}")
                    break
                else:
                    print(f"      ❌ Error at batch size {batch_size}: {e}")
                    break
            except Exception as e:
                print(f"      ❌ Unexpected error: {e}")
                break

        return batch_performance

    def run_batch_analysis(self, models_dict, base_prompt, max_batch_size=16):
        """Run batch analysis across all models"""
        all_results = {}
        batch_sizes = [1, 2, 4, 8, 16]

        for model_name, model_data in models_dict.items():
            short_name = model_name.split('/')[-1]

            print(f"\n🎯 Analyzing batch processing for {short_name}")

            batch_results = self.test_batch_processing(
                model_data['model'],
                model_data['tokenizer'],
                base_prompt,
                batch_sizes=[bs for bs in batch_sizes if bs <= max_batch_size]
            )

            if batch_results:
                all_results[model_name] = {
                    'model_info': {
                        'name': short_name,
                        'parameters': model_data['parameters'],
                        'size_category': model_data['size_category'],
                        'architecture': model_data['architecture']
                    },
                    'batch_results': batch_results
                }

        self.batch_results = all_results
        return all_results

# Run batch processing analysis
batch_analyzer = BatchProcessingAnalyzer(production_monitor)
base_prompt = "Explain the benefits of artificial intelligence in healthcare:"

if models_dict:
    batch_results = batch_analyzer.run_batch_analysis(models_dict, base_prompt, max_batch_size=8)

    print(f"\n✅ Batch processing analysis completed: {len(batch_results)} models")
```

```python
# Cell 20: Correct WorkloadSimulator for LoRA Compatibility
print("🛠️ CORRECTING WorkloadSimulator FOR LoRA COMPATIBILITY")
print("=" * 50)

# Redefine the WorkloadSimulator class with the corrected generate call
class WorkloadSimulator:
    def __init__(self, monitor):
        self.monitor = monitor
        self.workload_results = {}

    def simulate_workload_scenario(self, scenario_name, model, tokenizer, **kwargs):
        """Simulate different real-world workload scenarios"""

        print(f"\n🔄 Simulating: {scenario_name}")

        scenarios = {
            'chat_conversation': {
                'prompts': [
                    "Hello, how are you today?",
                    "What can you tell me about artificial intelligence?",
                    "How does machine learning work?",
                    "What are the benefits of renewable energy?",
                    "Can you explain quantum computing?"
                ],
                'max_length': 100,
                'temperature': 0.7
            },
            'content_generation': {
                'prompts': [
                    "Write a short story about a robot learning to paint:",
                    "Explain the concept of neural networks in simple terms:",
                    "Describe the impact of climate change on coastal cities:",
                    "Create a product description for a new smartphone:"
                ],
                'max_length': 150,
                'temperature': 0.8
            },
            'technical_qa': {
                'prompts': [
                    "What is the difference between supervised and unsupervised learning?",
                    "How does attention mechanism work in transformers?",
                    "Explain the concept of backpropagation in neural networks:",
                    "What are the advantages of using GPU for deep learning?"
                ],
                'max_length': 120,
                'temperature': 0.6
            }
        }

        scenario_config = scenarios.get(scenario_name, scenarios['chat_conversation'])

        performance_metrics = {
            'scenario': scenario_name,
            'total_requests': len(scenario_config['prompts']),
            'inference_times': [],
            'memory_usage': [],
            'throughput': [],
            'generated_lengths': []
        }

        for i, prompt in enumerate(scenario_config['prompts']):
            # Clear cache between requests (simulating real production)
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
                torch.cuda.synchronize()

            memory_before = torch.cuda.memory_allocated() / 1e9
            start_time = time.time()

            inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512).to(model.device)

            with torch.no_grad():
                # CORRECTED: Pass input_ids as a keyword argument
                outputs = model.generate(
                    input_ids=inputs.input_ids,
                    max_length=scenario_config['max_length'],
                    temperature=scenario_config['temperature'],
                    do_sample=True,
                    pad_token_id=tokenizer.eos_token_id,
                    attention_mask=inputs.attention_mask
                )

            if torch.cuda.is_available():
                torch.cuda.synchronize()

            end_time = time.time()
            memory_after = torch.cuda.memory_allocated() / 1e9

            inference_time = end_time - start_time
            memory_used = memory_after - memory_before

            # Calculate tokens and throughput
            generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
            input_tokens = len(tokenizer.encode(prompt))
            output_tokens = len(tokenizer.encode(generated_text))
            total_tokens = output_tokens - input_tokens
            tokens_per_sec = total_tokens / inference_time if inference_time > 0 else 0

            performance_metrics['inference_times'].append(inference_time)
            performance_metrics['memory_usage'].append(memory_used)
            performance_metrics['throughput'].append(tokens_per_sec)
            performance_metrics['generated_lengths'].append(len(generated_text.split()))

            # Record metrics
            self.monitor.collect_comprehensive_metrics(
                f"{scenario_name}_request_{i}",
                model.config.name_or_path if hasattr(model.config, 'name_or_path') else "unknown",
                inference_time
            )

            if i < 2:  # Print first few for verification
                print(f"   Request {i+1}: {inference_time:.3f}s, {tokens_per_sec:.1f} tok/s")


        # Calculate summary statistics
        performance_metrics.update({
            'avg_inference_time': np.mean(performance_metrics['inference_times']),
            'std_inference_time': np.std(performance_metrics['inference_times']),
            'avg_throughput': np.mean(performance_metrics['throughput']),
            'avg_memory_usage': np.mean(performance_metrics['memory_usage']),
            'total_tokens_generated': sum([len(tokenizer.encode(p)) for p in scenario_config['prompts']])
        })

        return performance_metrics


    def run_comprehensive_workloads(self, models_dict, scenarios=None):
        """Run multiple workload scenarios across all models"""
        if scenarios is None:
            scenarios = ['chat_conversation', 'content_generation', 'technical_qa']

        all_results = []

        for model_name, model_data in models_dict.items():
            short_name = model_name.split('/')[-1]
            print(f"\n🎯 Testing {short_name} across {len(scenarios)} scenarios")

            model_results = {
                'model': short_name,
                'parameters': model_data['parameters'],
                'size_category': model_data['size_category'],
                'architecture': model_data['architecture'],
                'scenarios': {}
            }

            for scenario in scenarios:
                try:
                    scenario_results = self.simulate_workload_scenario(
                        scenario,
                        model_data['model'],
                        model_data['tokenizer']
                    )
                    model_results['scenarios'][scenario] = scenario_results

                    print(f"   {scenario}: {scenario_results['avg_throughput']:.1f} tok/s")

                except Exception as e:
                    print(f"   ❌ {scenario} failed: {e}")
                    continue

            all_results.append(model_results)

        self.workload_results = all_results
        return all_results

# Re-initialize workload simulator with the corrected class definition
workload_simulator = WorkloadSimulator(production_monitor)

# Rerun workload simulation for the LoRA model
lora_models_dict_for_simulation = {
    'lora_model': {
        'model': lora_model,
        'tokenizer': base_tokenizer,
        'parameters': models_dict[selected_model_name_for_lora]['parameters'],
        'size_category': models_dict[selected_model_name_for_lora]['size_category'],
        'architecture': models_dict[selected_model_name_for_lora]['architecture']
    }
}

lora_workload_results = workload_simulator.run_comprehensive_workloads(lora_models_dict_for_simulation)

print("\n✅ Workload simulation for LoRA model completed.")
```
